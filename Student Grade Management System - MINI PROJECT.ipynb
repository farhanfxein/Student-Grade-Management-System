```python
"""
MINI PROJECT1: Student Grade Management System

Objective: Build a system to store, update, and retrieve student grades.

Task:

• Create a dictionary where student names are keys and their grades are values.

• Allow users to:
Add a new student.
Update an existing grade.
Retrieve a student’s grade.
Display all students.

Expected Output:
1. Add Student
2. Update Grade
3. Get Grade
4. Display All

5. Exit Enter choice: 1
Enter student name: Alice
Enter grade: A
Student added successfully.
```

```python
students = {}
```

```python
def add_student():
    name = input("Enter student fullname:")
    grade = input("Enter grade:")
    
    students[name] = grade
    print(f"Student added successfully")
```

```python
def update_grade():
    name = input("Enter student fullname:")
    
    if name in students:
        grade = input("Enter new grade:")
        students[name] = grade
        print(f"Grade updated successfully")
    else:
        print(f"Student not found")
```

```python
def get_grade():
    name = input("Enter student fullname: ")
    
    print(f"{name}'s grade: {students.get(name, 'Not found')}")
```

```python
def display_all():
    for name, grade in students.items():
        print(f"{name} : {grade}")
```

```python
while True:
    print("\n1. Add Student\n2. Update grade\n3. Get Grade\n4. Display All\n5. Exit")

    choice = input(f"Enter choice number: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        update_grade()
    elif choice == "3":
        get_grade()
    elif choice ==  "4":
        display_all()
    elif choice == "5":
        print("Exit Successful")
        break
    else:
        print(f"Invalid choice, Try again.")
```
